# Comparing Data at LWRT and Vanda

This code exists to compare data outputs at the Onyx @ LWRT gage and the Onyx @ Vanda gage to determine changes in the stream as it moves over the 20 miles of Wright Valley

```python
# Package Import
import datetime as dt
import matplotlib.dates as mdates
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import requests
from dotenv import load_dotenv
import os

myEDIAccessKey="fXTLLFP3rSJT-aTDY0ZqHJFTc6w"

# Importing data set from MCM online repository
VANDA_DF=pd.read_csv('https://pasta.lternet.edu/package/data/eml/knb-lter-mcm/9022/14/b383b37024d09bf2c6466d8bc0119282'.strip()+'?key='+myEDIAccessKey, 
                      dtype={'dischg_com':'str', 'comments':'str','water_temp_qlty':'str','conductivity_qlty':'str'})

LWRT_DF = pd.read_csv('https://pasta.lternet.edu/package/data/eml/knb-lter-mcm/9021/25/d883bf3e67b517ef7528bb0d23a14d24'.strip()+'?key='+myEDIAccessKey,
                      dtype={'dischg_com':'str', 'comments':'str','water_temp_qlty':'str','conductivity_qlty':'str'})


# Formatting the data set so that the 'date_time' column is in the proper format and 
VANDA_DF['date_time'] = pd.to_datetime(VANDA_DF['date_time'], format='mixed')
VANDA_DF['date_time'] = VANDA_DF['date_time'].apply(lambda x: x - pd.DateOffset(years=100) if x.year >= 2069 else x)

LWRT_DF['date_time'] = pd.to_datetime(LWRT_DF['date_time'], format='mixed')
LWRT_DF['date_time'] = LWRT_DF['date_time'].apply(lambda x: x - pd.DateOffset(years=100) if x.year >= 2069 else x)
```

```python
cols = ['date_time', 'discharge_ls', 'spcond_uscm']

lwrt = (
    LWRT_DF[cols]
    .rename(columns={
        'discharge_ls': 'Q_LWRT',
        'spcond_uscm': 'SPC_LWRT'
    })
    .copy()
)

vanda = (
    VANDA_DF[cols]
    .rename(columns={
        'discharge_ls': 'Q_VANDA',
        'spcond_uscm': 'SPC_VANDA'
    })
    .copy()
)

paired = pd.merge(
    lwrt,
    vanda,
    on='date_time',
    how='inner'
).sort_values('date_time')

paired['year'] = (
    paired['date_time'].dt.year
    + (paired['date_time'].dt.month >= 10).astype(int)
)
paired.head()
```

```python
year = 2020

season = paired[paired['year'] == year]

fig, ax = plt.subplots(figsize=(14, 5))

ax.plot(
    season['date_time'],
    season['Q_LWRT'],
    label='LWRT',
    linewidth=1.5
)

ax.plot(
    season['date_time'],
    season['Q_VANDA'],
    label='Vanda',
    linewidth=1.5
)

ax.set_title(f'Onyx River Discharge — {year}')
ax.set_ylabel('Discharge')
ax.legend()
ax.grid(alpha=0.2)

plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(6, 6))

ax.scatter(
    paired['Q_LWRT'],
    paired['Q_VANDA'],
    s=4,
    alpha=0.25
)

lims = [
    0,
    np.nanmax([
        paired['Q_LWRT'].max(),
        paired['Q_VANDA'].max()
    ])
]

ax.plot(lims, lims, '--', linewidth=1, color = 'black')

ax.set_xlim(lims)
ax.set_ylim(lims)

ax.set_xlabel('LWRT Discharge')
ax.set_ylabel('Vanda Discharge')
ax.set_title('Onyx River: Downstream Discharge Transformation')

plt.tight_layout()
plt.show()
```

```python
def discharge_lag(df, max_hours=48):
    
    x = df[['date_time', 'Q_LWRT', 'Q_VANDA']].dropna().copy()

    # Standardize so correlation reflects hydrograph shape
    lwrt = (
        (x['Q_LWRT'] - x['Q_LWRT'].mean()) /
        x['Q_LWRT'].std()
    )

    vanda = (
        (x['Q_VANDA'] - x['Q_VANDA'].mean()) /
        x['Q_VANDA'].std()
    )

    # Four 15-minute observations per hour
    max_lag = int(max_hours * 4)

    lags = np.arange(-max_lag, max_lag + 1)

    correlations = []

    for lag in lags:
        correlations.append(
            lwrt.corr(vanda.shift(-lag))
        )

    result = pd.DataFrame({
        'lag_steps': lags,
        'lag_hours': lags / 4,
        'correlation': correlations
    })

    return result
```

```python
lag_result = discharge_lag(paired, max_hours=48)

best = lag_result.loc[
    lag_result['correlation'].idxmax()
]

print(best)
```

```python
fig, ax = plt.subplots(figsize=(8, 4))

ax.plot(
    lag_result['lag_hours'],
    lag_result['correlation']
)

ax.axvline(
    best['lag_hours'],
    linestyle='--',
    label=f"Maximum correlation: {best['lag_hours']:.2f} h", color = 'black'
)

ax.set_xlabel('Lag (hours)')
ax.set_ylabel('Correlation')
ax.set_title('LWRT → Vanda Discharge Lag')
ax.legend()

plt.tight_layout()
plt.show()
```

```python
from scipy.signal import find_peaks

year = 2020
season = paired[paired['year'] == year].copy()

# Smooth slightly to avoid identifying tiny 15-min fluctuations
season['LWRT_smooth'] = (
    season['Q_LWRT']
    .rolling(4, center=True)
    .mean()
)

season['VANDA_smooth'] = (
    season['Q_VANDA']
    .rolling(4, center=True)
    .mean()
)

lwrt_peaks, _ = find_peaks(
    season['LWRT_smooth'].fillna(0),
    prominence=season['LWRT_smooth'].std()
)

vanda_peaks, _ = find_peaks(
    season['VANDA_smooth'].fillna(0),
    prominence=season['VANDA_smooth'].std()
)
```

```python
year = 2020
season = paired[paired['year'] == year]

fig, (ax1, ax2) = plt.subplots(
    2, 1,
    figsize=(14, 8),
    sharex=True
)

# Discharge
ax1.plot(
    season['date_time'],
    season['Q_LWRT'],
    label='LWRT'
)

ax1.plot(
    season['date_time'],
    season['Q_VANDA'],
    label='Vanda'
)

ax1.set_ylabel('Discharge')
ax1.legend()
ax1.grid(alpha=0.2)


# SPC
ax2.plot(
    season['date_time'],
    season['SPC_LWRT'],
    label='LWRT'
)

ax2.plot(
    season['date_time'],
    season['SPC_VANDA'],
    label='Vanda'
)

ax2.set_ylabel('Specific Conductance')
ax2.set_xlabel('Date')
ax2.legend()
ax2.grid(alpha=0.2)

fig.suptitle(f'Onyx River — {year}')

plt.tight_layout()
plt.show()
```

```python
# Make a copy so we don't modify the original dataframe
clim = paired.copy()

# Create a month-day-time key
clim['month_day_time'] = clim['date_time'].dt.strftime('%m-%d %H:%M')

# Calculate the mean across all years for each 15-minute timestamp
clim_mean = (
    clim
    .groupby('month_day_time', as_index=False)
    .agg({
        'Q_LWRT': 'mean',
        'Q_VANDA': 'mean',
        'SPC_LWRT': 'mean',
        'SPC_VANDA': 'mean'
    })
)

# Convert the grouping key back to a datetime for plotting.
# The year 2000 is arbitrary and is only being used as a plotting year.
clim_mean['plot_time'] = pd.to_datetime(
    '2000-' + clim_mean['month_day_time'],
    format='%Y-%m-%d %H:%M'
)

# Sort chronologically
clim_mean = clim_mean.sort_values('plot_time')
```

```python
fig, (ax1, ax2) = plt.subplots(
    2, 1,
    figsize=(14, 8),
    sharex=True
)

# -------------------------
# Mean discharge
# -------------------------
ax1.plot(
    clim_mean['plot_time'],
    clim_mean['Q_LWRT'],
    label='LWRT',
    linewidth=1.5
)

ax1.plot(
    clim_mean['plot_time'],
    clim_mean['Q_VANDA'],
    label='Vanda',
    linewidth=1.5
)

ax1.set_ylabel('Mean Discharge')
ax1.set_title('Mean 15-Minute Discharge')
ax1.legend()
ax1.grid(alpha=0.2)


# -------------------------
# Mean SPC
# -------------------------
ax2.plot(
    clim_mean['plot_time'],
    clim_mean['SPC_LWRT'],
    label='LWRT',
    linewidth=1.5
)

ax2.plot(
    clim_mean['plot_time'],
    clim_mean['SPC_VANDA'],
    label='Vanda',
    linewidth=1.5
)

ax2.set_ylabel('Mean Specific Conductance')
ax2.set_xlabel('Month')
ax2.set_title('Mean 15-Minute Specific Conductance')
ax2.legend()
ax2.grid(alpha=0.2)


# Format x-axis
ax2.xaxis.set_major_locator(
    mdates.MonthLocator()
)
ax2.xaxis.set_major_formatter(
    mdates.DateFormatter('%b')
)

fig.suptitle(
    'Onyx River — Mean 15-Minute Seasonal Conditions Across All Years',
    fontsize=14
)

plt.tight_layout()
plt.show()
```
